### Доллары в Рубли

```python
df["Пшеница 12,5% FOB Ново, руб/т"] = df["Пшеница 12,5% FOB Ново, $/т"] * df["USDRUB"]
df = df.drop("Пшеница 12,5% FOB Ново, $/т", axis=1)
```

### Кокосовое масло, табак, какао, пальмовое



```python
(df["Кокосовое масло (USD) "] * df["USDRUB"] - df["Кокосовое масло "])
```

По какой-то непонятной причине ненулевыми являются только последние  5 значение, скорее всего `Кокосовое масло (USD) ` было получено просто домножением на курс `USDRUB` но при этом где-то в концен ошибка

```python
(df["Какао-бобы (USD)"] * df["USDRUB"] - df["Какао-бобы"])
```

```python
(df["Табак (USD)"] * df["USDRUB"] - df["Табак"])
```

Видим что по какой-то причине ненулевыми являются только последние 5 значений. Переведем их в рубли, но оставим назавание колонки таким же

```python
df["Кокосовое масло (USD) "] = df["Кокосовое масло (USD) "] * df["USDRUB"]

df["Пальмовое масло (USD)"] = df["Пальмовое масло (USD)"] * df["USDRUB"]

df["Какао-бобы (USD)"] = df["Какао-бобы (USD)"] * df["USDRUB"]

df["Табак (USD)"] = df["Табак (USD)"] * df["USDRUB"]
```
